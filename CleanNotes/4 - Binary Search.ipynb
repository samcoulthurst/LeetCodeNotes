## Reference

### The plain algorithm

Halve a **sorted** search space each step -> O(log n) time, O(1) space.

1. `l, r = 0, len(nums) - 1` - the range is inclusive on both ends.
2. While `l <= r`: take `m = l + (r - l) // 2`.
3. Too big -> `r = m - 1`. Too small -> `l = m + 1`. Equal -> done.
4. Loop exits with `l > r` -> not found.

`l + (r - l) // 2` instead of `(l + r) // 2` avoids overflow (irrelevant in Python,
expected in C++/Java).

### The two conventions

Use `+/- 1` when the test **rules `m` out** (you compared it to the target):

```python
while l <= r:                  # l == r still holds an untested candidate
    m = (l + r) // 2
    if   nums[m] < target: l = m + 1
    elif nums[m] > target: r = m - 1
    else:                  return m
```

Use `r = m` when `m` **might be the answer** (smallest index satisfying X, minimums):

```python
while l < r:                   # l == r means found
    m = (l + r) // 2
    if ok(m): r = m            # keep m
    else:     l = m + 1        # m ruled out
return l
```

## Binary Search (LC 704)

````python
def search(nums, target):
    l, r = 0, len(nums) - 1

    while l <= r:
        m = l + (r - l) // 2

        if nums[m] > target:
            r = m - 1
        elif nums[m] < target:
            l = m + 1
        else:
            return m

    return -1
````

## Koko Eating Bananas (LC 875)

````python
import math

def minEatingSpeed(piles, h):
    l, r = 1, max(piles)   # candidate speeds
    res = r

    while l <= r:
        k = (l + r) // 2

        hours = 0
        for p in piles:
            hours += math.ceil(p / k)

        if hours <= h:     # fast enough - try slower
            res = k
            r = k - 1
        else:
            l = k + 1

    return res
````

## Find Minimum in Rotated Sorted Array (LC 153)

````python
def findMin(nums):
    l, r = 0, len(nums) - 1
    res = nums[0]

    while l <= r:
        if nums[l] < nums[r]:      # this stretch is already sorted
            res = min(res, nums[l])
            break

        m = (l + r) // 2
        res = min(res, nums[m])

        if nums[m] >= nums[l]:     # left half sorted -> pivot is right
            l = m + 1
        else:
            r = m - 1

    return res
````
