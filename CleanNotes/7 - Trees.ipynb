# Trees

**Reference**
- The `TreeNode` class
- DFS - depth first (recursion / stack)
- BFS - breadth first (queue)

**Problems**
- LC 226 - Invert Binary Tree
- LC 104 - Maximum Depth of Binary Tree
- LC 102 - Binary Tree Level Order Traversal

## Reference

Both traversals visit every node once -> O(n) time. The difference is the extra space
and the order you see nodes in.

| | Order | Uses | Space |
|---|---|---|---|
| **DFS** | down one branch to the bottom, then back up | recursion (implicit stack) | O(h) - height |
| **BFS** | level by level, left to right | `deque` queue | O(w) - widest level |

**DFS** - reach for it by default. Base case is always `if not node: return ...`, then
combine the results of `node.left` and `node.right`. Three orders, by *where* you touch
the node: preorder (copying/building a tree), inorder (a BST comes out sorted),
postorder (need the children's answers first: depth, diameter).

**BFS** - use when the answer depends on levels (level order, right side view, min
depth, shortest path). Pop a whole level at a time by snapshotting `len(q)` first.

Code for both below.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build(vals):
    """LeetCode's level-order list (None = missing) -> root."""
    if not vals:
        return None

    root = TreeNode(vals[0])
    q, i = [root], 1
    for node in q:
        for side in ("left", "right"):
            if i < len(vals):
                if vals[i] is not None:
                    child = TreeNode(vals[i])
                    setattr(node, side, child)
                    q.append(child)
                i += 1
    return root
```

## DFS and BFS

```python
from collections import deque


# --- DFS: recursive ---------------------------------------------------------

def dfs(node, res):
    if not node:
        return res

    res.append(node.val)      # PREorder  - before the children
    dfs(node.left, res)
    # res.append(node.val)    # INorder   - between them (sorted, for a BST)
    dfs(node.right, res)
    # res.append(node.val)    # POSTorder - after both children

    return res


# --- DFS: iterative (explicit stack, preorder) ------------------------------

def dfs_iter(root):
    res, stack = [], [root] if root else []

    while stack:
        node = stack.pop()
        res.append(node.val)

        if node.right:        # right pushed first -> left pops first
            stack.append(node.right)
        if node.left:
            stack.append(node.left)

    return res


# --- BFS: queue, level by level ---------------------------------------------

def bfs(root):
    res, q = [], deque([root] if root else [])

    while q:
        for _ in range(len(q)):   # snapshot: exactly one level
            node = q.popleft()
            res.append(node.val)

            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)

    return res
```

## Invert Binary Tree (LC 226)

```python
def invertTree(root):
    if not root:
        return None

    root.left, root.right = root.right, root.left

    invertTree(root.left)
    invertTree(root.right)

    return root
```

## Maximum Depth of Binary Tree (LC 104)

```python
def maxDepth(root):
    if not root:
        return 0

    return 1 + max(maxDepth(root.left), maxDepth(root.right))
```

## Binary Tree Level Order Traversal (LC 102)

```python
from collections import deque

def levelOrder(root):
    res = []
    q = deque([root] if root else [])

    while q:
        level = []
        for _ in range(len(q)):     # snapshot: exactly this level
            node = q.popleft()
            level.append(node.val)

            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)

        res.append(level)

    return res
```
