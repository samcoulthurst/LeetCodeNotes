## Reference: `defaultdict` and `dict` complexity

### `defaultdict`

`from collections import defaultdict`

A `dict` subclass that takes a **factory function** and calls it to create a value
whenever a missing key is looked up - so you never have to check "is the key there yet?".

```python
d = defaultdict(list)   # missing key -> []     d[k].append(x)
d = defaultdict(int)    # missing key -> 0      d[k] += 1
d = defaultdict(set)    # missing key -> set()  d[k].add(x)
```

Plain-`dict` equivalents:

```python
d[k] = d.get(k, 0) + 1        # counting
d.setdefault(k, []).append(x) # grouping
```

### `dict` operations and time complexity

| Operation | Example | Average | Worst |
|---|---|---|---|
| Lookup | `d[k]`, `d.get(k)` | O(1) | O(n) |
| Membership | `k in d` | O(1) | O(n) |
| Insert / update | `d[k] = v`, `d.setdefault(k, v)` | O(1) | O(n) |
| Delete | `del d[k]`, `d.pop(k)` | O(1) | O(n) |
| Size | `len(d)` | O(1) | O(1) |
| Iterate | `for k in d`, `.keys()`, `.values()`, `.items()` | O(n) | O(n) |
| Copy / build from iterable | `dict(d)`, `d.copy()`, `d.update(other)` | O(n) | O(n) |

Worst case is O(n) only when every key collides in the hash table
so treat dict/set access as O(1) in interviews.


## Two Sum (LC 1)

````python
def twoSum(nums, target):
    seen = {}
    for i, num in enumerate(nums):

        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return []
````

## Group Anagrams (LC 49)

````python
from collections import defaultdict

def groupAnagrams(strs):
    res = defaultdict(list)
    for s in strs:
        count = [0] * 26
        for c in s:
            count[ord(c) - ord('a')] += 1
        res[tuple(count)].append(s)
    return list(res.values())
````

## Longest Consecutive Sequence (LC 128)

````python
def longestConsecutive(nums):
    num_set = set(nums)
    longest = 0

    for n in num_set:
        # only start counting from the beginning of a sequence
        if (n - 1) not in num_set:
            length = 1
            while (n + length) in num_set:
                length += 1
            longest = max(longest, length)

    return longest
````
