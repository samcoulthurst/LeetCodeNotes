## Reference

### A stack is just a list

LIFO - push and pop the same end. Python needs no stack class:

```python
stack = []
stack.append(x)   # push   O(1)
stack.pop()       # pop    O(1)
stack[-1]         # peek   O(1)
if not stack:     # empty?
```

(Never `pop(0)` - that's O(n). For a queue use `collections.deque`.)

### Monotonic stack

A stack kept sorted (increasing or decreasing) by popping everything that breaks the
order before pushing. Each element is pushed and popped once -> O(n).

```python
for i, x in enumerate(arr):
    while stack and breaks_order(stack[-1], x):
        j = stack.pop()      # x is the answer for j
    stack.append(i)
```

Use it for "next/previous greater/smaller element" questions - decreasing stack for
next greater, increasing stack for next smaller.

## Valid Parentheses (LC 20)

````python
def isValid(s):
    stack = []
    closeToOpen = {")": "(", "]": "[", "}": "{"}

    for c in s:
        if c in closeToOpen:
            if stack and stack[-1] == closeToOpen[c]:
                stack.pop()
            else:
                return False
        else:
            stack.append(c)

    return not stack
````

## Daily Temperatures (LC 739)

````python
def dailyTemperatures(temperatures):
    res = [0] * len(temperatures)
    stack = []  # decreasing: (temp, index)

    for i, t in enumerate(temperatures):
        while stack and t > stack[-1][0]:
            stackT, stackInd = stack.pop()
            res[stackInd] = i - stackInd
        stack.append((t, i))

    return res
````
