## Reference

### Converging pointers (left / right)

Start at both ends and walk inwards. Use when the array is **sorted**, or when the
answer depends on a **pair** taken from opposite ends.


### Fast and slow pointers

Both start at the front, moving at different speeds (or one only advances on a
condition). Use for **linked lists** and **in-place array rewrites** - no sorting needed.

## Two Sum II - Input Array Is Sorted (LC 167)

```python
def twoSum(numbers, target):
    l, r = 0, len(numbers) - 1

    while l < r:
        curSum = numbers[l] + numbers[r]

        if curSum > target:
            r -= 1
        elif curSum < target:
            l += 1
        else:
            return [l + 1, r + 1]  # 1-indexed

    return []
```

## Container With Most Water (LC 11)

```python
def maxArea(heights):
    l, r = 0, len(heights) - 1
    res = 0

    while l < r:
        area = (r - l) * min(heights[l], heights[r])
        res = max(res, area)

        # move the shorter wall - the taller one can't improve
        if heights[l] <= heights[r]:
            l += 1
        else:
            r -= 1

    return res
```
