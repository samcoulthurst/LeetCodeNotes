```python
def findSplit(nums):
    l = 0 
    r = len(nums) - 1
    while l < r:
        m = (l + r) // 2
        if nums[m] > nums[r]:
            l = m + 1 
        elif nums[m] < nums[l]:
            r = m
        else:
            return l
    return l   
```

```python
findSplit(nums)
```

```python
def BinarySearch(nums,target):
    l = 0
    r = len(nums) - 1
    while l <= r:
        m = (l + r) // 2
        if target < nums[m]:
            r = m - 1
        elif target > nums[m]:
            l = m + 1
        else:
            return m
    return -1
```

```python
def searchInRotated(nums, target):
    splitIDX = findSplit(nums)
    print(splitIDX)
    half1 = nums[0:splitIDX]
    half2 = nums[splitIDX:] 
    b1 = BinarySearch(half1,target)
    b2 = BinarySearch(half2,target)
    print(b1,b2)
    if b2 == -1:
        return max(b1, b2)
    else:
        return max(b1, splitIDX + b2)
```

```python
searchInRotated(nums=[3,5,1], target=3)
```
