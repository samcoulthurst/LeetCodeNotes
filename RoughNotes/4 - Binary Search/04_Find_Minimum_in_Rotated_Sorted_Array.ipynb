# Find Minimum in Rotated Sorted Array (LeetCode 153)

```python
nums = [4,5,0,1,2,3]
```

```python
def findMin(nums):
    l = 0
    r = len(nums) - 1
    while l <= r:
        m = (l + r) // 2
        print(l,m,r)
        if nums[l] < nums[m] and nums[m] < nums[r]:
            return nums[l] 
        elif nums[r] < nums[l] and nums[l] < nums[m]:
            l = m + 1
        elif nums[m] < nums[r] and nums[r] < nums[l]:
            r = m
        if l == r or l==m:
            return min(nums[l],nums[r]) 
```

```python
x = findMin(nums)
print(x)
```
