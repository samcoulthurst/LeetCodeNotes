# Search a 2D Matrix (LeetCode 74)

## Notes

```python
matrix = [[1,2,4,8],[10,11,12,13],[14,20,30,40]]
target = 10
```

```python
def BinarySearch(array, target):
    l = 0
    r = len(array) - 1
    while l <= r:
        m = (l+r) // 2
        if array[m] < target:
            l = m + 1
        elif array[m] > target:
            r = m - 1
        else:
            return m
    return -1
```

```python
rows = len(matrix)
top = 0
bottom = rows - 1

while top <= bottom:
    m = (top + bottom) // 2
    print(m)
    if target > matrix[m][-1]:
        top = m + 1
    elif target < matrix[m][0]:
        bottom = m - 1
    else:
        row = m
        break
    
```

```python
print(top)
print(bottom)
print(row)
```

```python
col = BinarySearch(matrix[row], target)
```

```python
coords = (row, col)
print(coords)
```
