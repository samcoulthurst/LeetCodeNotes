# Best Time to Buy and Sell Stock (LeetCode 121)

## Notes

```python
prices = [10,1,5,6,7,1]
```

```python
res = []
l = len(prices)
for i in range(l):
    res_price = []
    for j in range(i,l):
        res_price.append(prices[j]-prices[i])
    res.append(res_price)

print(res)
print(max(max(res)))
```

```python
prices = [10,8,7,5,2]
```

```python
max_profit = 0
current_smallest = 99999999
l = len(prices)
for i in range(l-1):
    buy = prices[i]
    sell = prices[i+1]
    print(buy,sell)
    current_smallest = min(current_smallest, buy)
    print(f"current Smallest: {current_smallest}")
    profit = sell - current_smallest
    print(f"profit: {profit}")
    max_profit = max(profit, max_profit)
```
