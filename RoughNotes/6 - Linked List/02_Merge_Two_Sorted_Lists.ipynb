```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
def link(_list):
    head = ListNode(_list[0])
    curr = head
    for i in range(1,len(_list)):
        #print(_list[i])
        curr.next = ListNode(_list[i])
        curr = curr.next

    return head

def unlink(head):
    curr = head
    res = []
    while curr:
        res.append(curr.val)
        curr = curr.next
    return res
```

```python
list1 = ListNode(1)
list1.next = ListNode(2)
list1.next.next = ListNode(4)

list2 = ListNode(1)
list2.next = ListNode(3)
list2.next.next = ListNode(5)
```

```python
def mergeTwoLists(list1, list2):
    p1 = list1
    p2 = list2

    res = ListNode(None)
    curr = res

    while p1 or p2:
        
        if not p1:
            curr.next = ListNode(p2.val)
            curr = curr.next
            p2 = p2.next
        
        elif not p2:
            curr.next = ListNode(p1.val)
            curr = curr.next
            p1 = p1.next
        else:

            if p1.val < p2.val:
                curr.next = ListNode(p1.val)
                curr = curr.next
                p1 = p1.next
            else:
                curr.next = ListNode(p2.val)
                curr = curr.next
                p2 = p2.next
    return res
```
