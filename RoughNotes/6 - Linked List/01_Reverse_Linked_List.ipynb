# Reverse Linked List (LeetCode 206)

## Notes

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
x = ListNode(1)
x.next = ListNode(2)
x.next.next = ListNode(3)
x.next.next.next = ListNode(4)
print(x)
```

```python
def unlink(head):
    curr = head
    res = []
    while curr:
        res.append(curr.val)
        curr = curr.next
    return res
```

```python
y = unlink(x)
print(y)
```

```python
def link(_list):
    head = ListNode(_list[0])
    curr = head
    for i in range(1,len(_list)):
        #print(_list[i])
        curr.next = ListNode(_list[i])
        curr = curr.next

    return head
```

```python
z = link(y)
print(unlink(z))
```

```python
x = ListNode(1)
x.next = ListNode(2)
x.next.next = ListNode(3)
x.next.next.next = ListNode(4)
print(unlink(x))
```

```python
def reverseList(head):
    prev = None
    curr = head
    while curr:
        nxt = curr.next
        curr.next = prev

        prev = curr
        curr = nxt
    return prev
```

```python
r = reverseList(x)
print(unlink(r))
```
