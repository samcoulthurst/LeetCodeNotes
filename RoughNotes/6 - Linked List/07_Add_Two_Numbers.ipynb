```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def link(_list):
    head = ListNode(_list[0])
    curr = head
    for i in range(1,len(_list)):
        #print(_list[i])
        curr.next = ListNode(_list[i])
        curr = curr.next

    return head

def unlink(head):
    curr = head
    res = []
    while curr:
        res.append(curr.val)
        curr = curr.next
    return res
```
