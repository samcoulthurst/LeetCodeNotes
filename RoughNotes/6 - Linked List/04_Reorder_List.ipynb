```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
def link(_list):
    head = ListNode(_list[0])
    curr = head
    for i in range(1,len(_list)):
        #print(_list[i])
        curr.next = ListNode(_list[i])
        curr = curr.next

    return head

def unlink(head):
    curr = head
    res = []
    while curr:
        res.append(curr.val)
        curr = curr.next
    return res
```

```python
def reverseList(head):
    curr = head
    prev = None

    while curr:
        nxt = curr.next
        curr.next = prev

        prev = curr
        curr = nxt
    return prev
```

```python
def middle(head):
    slow = head
    fast = head

    res = ListNode(None)
    curr = res

    while fast and fast.next:
        curr.next = ListNode(slow.val)
        curr = curr.next

        slow = slow.next
        fast = fast.next.next

    # Add the middle node
    curr.next = ListNode(slow.val)

    return res.next
```

```python
def mergeTwoLists(list1, list2):
    dummy = node = ListNode()

    while list1 and list2:
        next1 = list1.next
        next2 = list2.next

        node.next = list1
        list1.next = list2

        node = list2
        list1 = next1
        list2 = next2

    node.next = list1 or list2

    return dummy.next
```

```python
def reorder_list(head):
    unrev = head
    rev = reverseList(head.next)

    middle_unrev = middle(unrev)
    middle_rev = middle(rev)

    return mergeTwoLists(middle_unrev,middle_rev)

     
```
