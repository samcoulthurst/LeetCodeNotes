```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if s is None or t is None:
            return False
        if len(s) != len(t):
            return False

        hash_s = {}
        hash_t = {}

        for i in range(len(s)):
            hash_s[s[i]] = hash_s.get(s[i],0) + 1
            hash_t[t[i]] = hash_t.get(t[i],0) + 1

        return hash_s == hash_t
```
