# Group Anagrams (LeetCode 49)

## Notes

```python
strs = ["act","pots","tops","cat","stop","hat"]
```

```python
from collections import defaultdict
```

```python
from collections import defaultdict

seen = defaultdict(list)

for word in strs:
    if "".join(sorted(word)) in seen:
        print('yes')
    seen["".join(sorted(word))].append(word)

print(list(seen.values()))
```
