```python
nums=[1,1,1,2,2,3]
k=2
```

```python
hash_map = {}
for num in nums:
    hash_map[num] = hash_map.get(num,0) + 1
print(hash_map)
sorted_hash = dict(sorted(hash_map.items(), key=lambda x: x[1], reverse=True))
print(sorted_hash)
print(list(sorted_hash.keys())[:k])
```
