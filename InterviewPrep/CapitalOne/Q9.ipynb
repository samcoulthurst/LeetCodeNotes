Given a sorted array arr and a target value, return the indices of two elements that sum to the target. If no such pair exists, return [-1, -1].

Example: arr = [1, 2, 3, 5, 8], target = 10 → [2, 4] (since arr[2] + arr[4] = 3 + 8 = 10)

```python
arr = [1, 2, 3, 5, 8]
target = 10
```

```python
left = 0
right = len(arr) - 1
while left < right:
    value = arr[left] + arr[right]
    if value < target:
        left += 1
    elif value > target:
        right -= 1
    else:
        print(arr[left], arr[right])
        break
```
