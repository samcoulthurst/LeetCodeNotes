Given a string s, return the character that appears most frequently. If there's a tie, return the lexicographically smallest character among the tied ones.

Example: s = "aababccc" → "a" (appears 3 times, same as "c", but "a" < "c" alphabetically)

```python
from collections import defaultdict
```

```python
s = "aababccc"
counts = defaultdict()
for letter in s:
    counts[letter] = counts.get(letter,0) + 1
```

```python
def highFreq(string):
    counts = {}
    for letter in string:
        counts[letter] = counts.get(letter, 0) + 1
    max_freq = max(counts.values())
    return min(c for c, f in counts.items() if f == max_freq)
```
