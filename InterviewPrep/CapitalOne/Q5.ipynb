Question 5 — Stack Problem: Valid Parentheses

Given a string s containing only the characters (, ), {, }, [, ], determine whether the brackets are properly balanced (every opening bracket is closed by the same type, in the correct order).

Example: "({[]})" → True, "({[})" → False

```python
lookup = {'[':']',
          '{':'}',
          '(':')'  }
```

```python
example = "({[]})"
stack = []
if len(example) % 2 != 0:
    print('False')
for bracket in example:
    if bracket in lookup: # if we are an opening bracket
        stack.append(bracket) # add to stack
        #print(stack)
    else: # we are a closing bracket
        #print(bracket, lookup[stack[-1]])
        if bracket != lookup[stack[-1]]:   # if we do not close the opening bracket of same type we are false
            print('False')
        stack.pop()

print('True')
```

```python
def validParantheses(string):
    stack = []
    if len(example) % 2 != 0:
        return False
    if string == "":
        return True
    
    for bracket in example:
        if bracket in lookup: # if we are an opening bracket
            stack.append(bracket) # add to stack
            print(stack)
        else: # we are a closing bracket
            #print(bracket, lookup[stack[-1]])
            
            if stack == [] or bracket != lookup[stack[-1]]:   # if we do not close the opening bracket of same type we are false
                return False
            stack.pop()
            print(stack)
    return not stack
```
