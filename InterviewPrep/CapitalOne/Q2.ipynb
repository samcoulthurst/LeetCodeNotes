Question 2 — String Pattern Matching: Anagram Groups

Given an array of strings strs, group the anagrams together. Return the groups in any order.

Example: ["eat","tea","tan","ate","nat","bat"] → [["eat","tea","ate"],["tan","nat"],["bat"]]

```python
words = ["eat","tea","tan","ate","nat","bat"]
```

```python
from collections import defaultdict
```

```python
uniques = {}
for word in words:
    sorted_word = "".join(sorted(word))
    if sorted_word not in uniques.keys():
        uniques[sorted_word] = [word]   
    else:
        uniques[sorted_word] += [word]
res = list(uniques.values())
res
```
