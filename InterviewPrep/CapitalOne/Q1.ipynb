Question 1 — Array Manipulation: Neighbor Sum

Given an array a, return a new array b where each element b[i] = a[i-1] + a[i] + a[i+1]. If i-1 or i+1 is out of bounds, treat that missing neighbor as 0.

Example: a = [4, 0, 1, -2, 3] → b = [4, 5, -1, 2, 1]

```python
a = [4, 0, 1, -2, 3]
```

```python
b = []
for i in range(len(a)):
    num = a[i]
    num += a[i-1] if i>0 else 0
    num += a[i+1] if i+1 < len(a) else 0
    b.append(num)
```

```python
b
```
