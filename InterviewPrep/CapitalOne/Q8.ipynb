Given an array arr and an integer k, return the maximum sum of any contiguous subarray of length exactly k.

Example: arr = [2, 1, 5, 1, 3, 2], k = 3 → 9 (the subarray [5, 1, 3])

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3
```

```python
current_arr = arr[:k]
best_arr = arr[:k]
print(best_arr)
```

```python
for i in range(k, len(arr)):
    print(f'current array is {current_arr} with total {sum(current_arr)}')
    print(f'the best array so far is {best_arr} with total {sum(best_arr)}')
    if sum(current_arr) > sum(best_arr):
        print('We have a new best')
        best_arr = current_arr
    current_arr.pop(0)
    current_arr.append(arr[i])
```

```python
window_sum = sum(arr[:k])
best = window_sum
best_arr = arr[:k]

for i in range(1, len(arr) - k +1):
    window_sum += arr[k + i-1] - arr[i-1]
    if window_sum > best:
        best = window_sum
        best_arr = arr[i:k+i]
```
